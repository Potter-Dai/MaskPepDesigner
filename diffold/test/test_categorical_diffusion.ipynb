```python
from diffold.diffusion.transition import SequenceCategoricalTransition

N = 10
cat_trans = SequenceCategoricalTransition("masking", num_classes=N)

import torch

batch_shape=(2,3)
seqlen = 5
aatype_t = torch.randint(0, N, size=(*batch_shape,seqlen))
gamma_t = torch.rand((*batch_shape, seqlen))
probs_s_hat = torch.rand((*batch_shape, seqlen, N))
gamma_s = (gamma_t + .2).clamp_max(1.)
```

```python
noise, prob = cat_trans.add_noise(aatype_t, gamma_t)
prob[..., -1], noise
```

```python
pred, poste = cat_trans.denoise(
    aatype_t,
    probs_s_hat=probs_s_hat,
    gamma_t=gamma_t,
    gamma_s=gamma_s
)
pred
```
